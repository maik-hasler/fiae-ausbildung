# Setup

Ihr könnt auf zwei Datenbanken arbeiten:

- **Variante A:** eure eigene Datenbank aus Teil 1. Tabellen- und Spaltennamen sind dann eure eigenen, passt die Abfragen also an. Ob ein Ergebnis stimmt, prüft ihr an euren eigenen Daten.
- **Variante B:** die Referenzdatenbank mit 5.000 Accounts und 15.000 Charakteren. Nur hier passen die Kontrollwerte.

**Empfehlung:** Löst jede Aufgabe zuerst auf eurer Datenbank und prüft sie dann auf der Referenz.

Die nächste Zelle baut die Referenzdatenbank einmalig als Datei `referenz.db` auf. Das dauert ein paar Sekunden.

```python
import sqlite3
from pathlib import Path

referenz = Path("referenz.db")
if not referenz.exists():
    with sqlite3.connect(referenz) as con:
        con.executescript(Path("../referenz/setup.sql").read_text(encoding="utf-8"))
print("Referenzdatenbank bereit:", referenz.resolve())
```

Jetzt verbindet ihr euch. Eine der beiden `%sql`-Zeilen ist mit `#` auskommentiert. Zum Umschalten das `#` von der einen Zeile entfernen und vor die andere setzen, dann die Zelle erneut ausführen.

`displaylimit = 20` sorgt dafür, dass bei großen Ergebnissen nur die ersten 20 Zeilen angezeigt werden.

```python
%load_ext sql
%config SqlMagic.displaylimit = 20

# Variante A: eigene Datenbank aus Teil 1
%sql sqlite:///../1-datenbank-entwerfen/meine_mmo.db

# Variante B: Referenzdatenbank
# %sql sqlite:///referenz.db
```
