```sql
SELECT g.name, SUM(p.gold) AS gold, ROUND(AVG(ps.value), 1) AS hp_schnitt
FROM guild g
JOIN player_guild pg ON pg.guild_id = g.id

JOIN player_stat ps  ON ps.player_id = p.id
```
