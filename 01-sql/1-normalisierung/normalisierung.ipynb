### Die Ausrüstung zerlegen
Eine Liste wie `Bogen, Heiltrank, Heiltrank` in einzelne Zeilen aufzuteilen, ist in SQL mühsam. Die Anweisungen unten erledigen das für euch, ihr müsst sie **nicht** im Detail verstehen. Kopiert sie in die Zelle darunter, sobald `charakter` existiert. Sie schneidet den Text an jedem Komma auf (`WITH RECURSIVE`) und zählt gleiche Gegenstände zusammen.

Merkt euch nur: **Kaputte Daten nachträglich zu reparieren, ist viel aufwendiger, als sie gleich richtig zu speichern.**

```sql
CREATE TABLE IF NOT EXISTS charakter_item (
    username TEXT NOT NULL REFERENCES charakter(username),
    item     TEXT NOT NULL,
    anzahl   INTEGER NOT NULL CHECK (anzahl > 0),
    PRIMARY KEY (username, item)
);
DELETE FROM charakter_item;

WITH RECURSIVE teile(username, rest, item) AS (
    SELECT username, ausruestung || ',', NULL
    FROM (SELECT DISTINCT username, ausruestung FROM gildenliste)
    UNION ALL
    SELECT username,
           SUBSTR(rest, INSTR(rest, ',') + 1),
           TRIM(SUBSTR(rest, 1, INSTR(rest, ',') - 1))
    FROM teile
    WHERE rest <> ''
)
INSERT INTO charakter_item (username, item, anzahl)
SELECT username, item, COUNT(*)
FROM teile
WHERE item IS NOT NULL
GROUP BY username, item;

SELECT * FROM charakter_item ORDER BY username, item;
```

# Teil C: Jetzt mit Fachbegriffen

Was ihr gerade gemacht habt, heißt **Normalisierung**. Die einzelnen Schritte haben Namen:

| Normalform | Regel | Das habt ihr gemacht |
|---|---|---|
| **1. Normalform (1NF)** | Jedes Feld enthält genau **einen** Wert, keine Listen. | Aus `ausruestung` wurde die Tabelle `charakter_item` mit einer Zeile pro Gegenstand. |
| **2. Normalform (2NF)** | 1NF, und jede Nicht-Schlüssel-Spalte hängt vom **ganzen** Schlüssel ab, nicht nur von einem Teil davon. | Der Schlüssel war `(username, gilde)`. `email` und `klasse` hingen nur am `username`, die Gildenbeschreibung nur an der `gilde`. Das kam in eigene Tabellen. In `mitgliedschaft` blieb nur, was beide Teile braucht: `rolle` und `beigetreten`. |
| **3. Normalform (3NF)** | 2NF, und keine Nicht-Schlüssel-Spalte hängt von einer **anderen Nicht-Schlüssel-Spalte** ab. | In `charakter` hing die Klassenbeschreibung an der `klasse`, nicht am `username`. Diese **transitive Abhängigkeit** (username → klasse → beschreibung) wurde zur Tabelle `klasse`. |

Ein Merksatz fasst alle drei zusammen: Jede Nicht-Schlüssel-Spalte hängt ab vom Schlüssel (1NF), vom **ganzen** Schlüssel (2NF) und von **nichts anderem als** dem Schlüssel (3NF).

Gut zu wissen:

- Die 2NF ist ein Thema, sobald ein Schlüssel aus **mehreren** Spalten besteht, auch ein fachlicher wie `(username, gilde)`.
- **Achtung, Falle:** Eine zusätzliche `id` als Primärschlüssel repariert nichts. Hätte die Gildenliste eine Spalte `id`, wären alle Anomalien trotzdem noch da: `(username, gilde)` bestimmt weiterhin eindeutig eine Zeile, und `email` hängt weiterhin nur am `username`.
- Es gibt noch strengere Normalformen (Boyce-Codd-Normalform, 4NF, 5NF). In der Praxis und in der Abschlussprüfung reicht meist die 3NF.

**Die Anomalien sind weg:**

| Anomalie | Vorher | Jetzt |
|---|---|---|
| Änderung | Gildenbeschreibung in 6 Zeilen | 1 Zeile in `gilde` |
| Einfügen | Gilde ohne Mitglied unmöglich | einfach `INSERT INTO gilde` |
| Löschen | letzte Mitgliedschaft weg, damit auch Gilde und Charakter weg | nur die Mitgliedschaft ist weg |
